# Stock Market Data Analysis

## 1. Data Collection

### Objective
Collect five years of historical stock market data
for selected Indian companies using Yahoo Finance.

### Stocks
- Reliance Industries
- TCS
- Infosys
- HDFC Bank
- ICICI Bank

```python
import pandas as pd
import numpy as np
import yfinance as yf

from pathlib import Path
```

```python
stocks = [
    "RELIANCE.NS",
    "TCS.NS",
    "INFY.NS",
    "HDFCBANK.NS",
    "ICICIBANK.NS"
]

print(stocks)
```

```python
all_data = []

for ticker in stocks:

    print(f"Downloading data for {ticker}...")

    df = yf.download(
        ticker,
        period="5y",
        interval="1d",
        auto_adjust=True,
        progress=False
    )

    if df.empty:
        print(f"No data found for {ticker}")
        continue

    # Handle MultiIndex columns
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)

    df = df.reset_index()

    df["Ticker"] = ticker

    all_data.append(df)

stock_data = pd.concat(
    all_data,
    ignore_index=True
)

print("\nData downloaded successfully!")
```

## 5. Exploratory Data Analysis (EDA)

Explore stock price movements, trading volume,
and historical trends using statistical analysis
and visualizations.

```python
# Sort data by company and date
cleaned_data = cleaned_data.sort_values(
    ["Ticker", "Date"]
).copy()

# Check the cleaned dataset
cleaned_data.head()
```

```python
cleaned_data = cleaned_data.reset_index(drop=True)

cleaned_data.head()
```

```python
plt.figure(figsize=(15, 12))

for i, ticker in enumerate(stocks, 1):

    company_data = cleaned_data[
        cleaned_data["Ticker"] == ticker
    ]

    plt.subplot(3, 2, i)

    plt.plot(
        company_data["Date"],
        company_data["Close"],
        label=ticker
    )

    plt.title(f"{ticker} - Historical Closing Price")
    plt.xlabel("Date")
    plt.ylabel("Adjusted Closing Price (₹)")
    plt.xticks(rotation=45)
    plt.grid(True, alpha=0.3)
    plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Calculate daily percentage returns for each stock

cleaned_data["Daily_Return"] = (
    cleaned_data.groupby("Ticker")["Close"]
    .pct_change() * 100
)

# Display the results
cleaned_data[
    ["Date", "Ticker", "Close", "Daily_Return"]
].head(10)
```

```python
# Count missing daily returns
print(
    "Missing daily returns:",
    cleaned_data["Daily_Return"].isnull().sum()
)

# Check missing returns company-wise
print(
    cleaned_data.groupby("Ticker")["Daily_Return"].apply(
        lambda x: x.isnull().sum()
    )
)
```

## 6. Risk and Return Analysis

Calculate statistical measures to understand
the returns and volatility of each stock.

```python
# Calculate risk and return statistics for each stock

risk_return = cleaned_data.groupby("Ticker")["Daily_Return"].agg(
    Mean_Return="mean",
    Volatility="std",
    Maximum_Return="max",
    Minimum_Return="min"
)

# Display results
risk_return
```

```python
# Calculate annualized volatility

risk_return["Annualized_Volatility"] = (
    risk_return["Volatility"] * np.sqrt(252)
)

# Display updated results
risk_return
```

```python
# Approximate annualized mean return

risk_return["Annualized_Mean_Return"] = (
    risk_return["Mean_Return"] * 252
)

risk_return
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    risk_return["Annualized_Volatility"],
    risk_return["Annualized_Mean_Return"],
    s=100
)

# Add company names beside each point
for ticker, row in risk_return.iterrows():

    plt.annotate(
        ticker,
        (
            row["Annualized_Volatility"],
            row["Annualized_Mean_Return"]
        ),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title("Risk vs Return Analysis")
plt.xlabel("Annualized Volatility (%)")
plt.ylabel("Approx. Annualized Mean Return (%)")

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Calculate growth of an initial investment of ₹100

cleaned_data["Growth_of_100"] = (
    cleaned_data.groupby("Ticker")["Daily_Return"]
    .transform(
        lambda returns: (
            1 + returns.fillna(0) / 100
        ).cumprod() * 100
    )
)

cleaned_data[
    ["Date", "Ticker", "Daily_Return", "Growth_of_100"]
].head(10)
```

```python
# Create a growth comparison chart

plt.figure(figsize=(15, 7))

for ticker in stocks:

    company_data = cleaned_data[
        cleaned_data["Ticker"] == ticker
    ]

    plt.plot(
        company_data["Date"],
        company_data["Growth_of_100"],
        label=ticker
    )

# Add reference line at ₹100
plt.axhline(
    y=100,
    linestyle="--",
    alpha=0.7,
    label="Initial Investment (₹100)"
)

plt.title("Growth of ₹100 - Stock Comparison")
plt.xlabel("Date")
plt.ylabel("Investment Value (₹)")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Calculate final investment value for each stock

final_performance = cleaned_data.groupby("Ticker").agg(
    Initial_Value=("Growth_of_100", "first"),
    Final_Value=("Growth_of_100", "last")
)

# Calculate total percentage return
final_performance["Total_Return_Percent"] = (
    (final_performance["Final_Value"] /
     final_performance["Initial_Value"] - 1) * 100
)

# Display results
final_performance.round(2)
```

```python
# Calculate the actual investment period for each stock

date_range = cleaned_data.groupby("Ticker")["Date"].agg(
    ["min", "max"]
)

period_years = (
    (date_range["max"] - date_range["min"]).dt.days / 365.25
)

# Add investment period to performance table
final_performance["Period_Years"] = period_years

# Calculate CAGR
final_performance["CAGR_Percent"] = (
    (
        final_performance["Final_Value"] /
        final_performance["Initial_Value"]
    ) ** (1 / final_performance["Period_Years"]) - 1
) * 100

# Display results
final_performance.round(2)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    risk_return["Annualized_Volatility"],
    final_performance["CAGR_Percent"],
    s=120
)

# Label each stock
for ticker in risk_return.index:

    plt.annotate(
        ticker,
        (
            risk_return.loc[ticker, "Annualized_Volatility"],
            final_performance.loc[ticker, "CAGR_Percent"]
        ),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.axhline(y=0, linestyle="--", alpha=0.6)

plt.title("CAGR vs Risk Analysis")
plt.xlabel("Annualized Volatility (%)")
plt.ylabel("CAGR (%)")

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Assume a 0% annual risk-free rate for this exercise

risk_free_rate = 0.0

# Calculate Sharpe Ratio
risk_return["Sharpe_Ratio"] = (
    (risk_return["Annualized_Mean_Return"] - risk_free_rate)
    / risk_return["Annualized_Volatility"]
)

# Display results
risk_return[
    ["Annualized_Mean_Return",
     "Annualized_Volatility",
     "Sharpe_Ratio"]
].round(3)
```

```python
# Combine risk and return statistics with investment performance

stock_report = risk_return.join(
    final_performance[
        [
            "Final_Value",
            "Total_Return_Percent",
            "CAGR_Percent"
        ]
    ])

# Sort companies by CAGR (highest first)
stock_report = stock_report.sort_values(
    by="CAGR_Percent",
    ascending=False
)

# Display the consolidated report
stock_report.round(2)
```

## 7. Stock Performance Dashboard

Visual comparison of historical returns, volatility,
and risk-adjusted performance across selected stocks.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. CAGR comparison
stock_report["CAGR_Percent"].plot(
    kind="bar",
    ax=axes[0],
    title="CAGR Comparison"
)

axes[0].set_ylabel("CAGR (%)")
axes[0].axhline(y=0, linestyle="--", alpha=0.5)
axes[0].tick_params(axis="x", rotation=45)

# 2. Annualized volatility
stock_report["Annualized_Volatility"].plot(
    kind="bar",
    ax=axes[1],
    title="Annualized Volatility"
)

axes[1].set_ylabel("Volatility (%)")
axes[1].tick_params(axis="x", rotation=45)

# 3. Sharpe Ratio
stock_report["Sharpe_Ratio"].plot(
    kind="bar",
    ax=axes[2],
    title="Sharpe Ratio"
)

axes[2].set_ylabel("Sharpe Ratio")
axes[2].axhline(y=0, linestyle="--", alpha=0.5)
axes[2].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```
